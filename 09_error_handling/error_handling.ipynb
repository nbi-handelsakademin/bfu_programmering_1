## Try - except

- try-block för att testa "farlig kod", dvs kod som kan ge error                                                             

- except-block - hanterar felet istället för att krascha programmet

- raise - kasta ett fel med specifik feltext

</div>


```python
numbers = list(range(10))

try:
  numbers[10]
except IndexError as err:
  print(err)

print("program fortsätter")
```

```python
age = input("Vad är din ålder?") # -5

try: 
    age = int(age)
    if not 0<=age<=125: 
        raise ValueError(f"Ålder måste vara mellan 0 och 125, inte {age}") 
    print(f"Du är {age} år gammal")
except ValueError as err:
    print(err)
```

Fråga flera gånger

```python
while True: 
    age = input("Vad är din ålder?") # -5

    try: 
        age = int(age)
        if not 0<=age<=125: 
            raise ValueError(f"Ålder måste vara mellan 0 och 125, inte {age}") 
        print(f"Du är {age} år gammal")
        break
    except ValueError as err:
        print(err)
```
