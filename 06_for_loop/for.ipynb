# for-sats

for loop används för att repetera kod, exempelvis loopa över sekvenser som listor. Den är framförallt bra när man vet hur många gånger i förväg som man vill upprepa, eller om det finns någon fixt sekvens eller lista att loopa över

## range() funktion

Används tillsammans med for för att loopa koden ett visst antal gånger

```python
for i in range(n): 
    print(i)
```
i startas från 0 och slutar på n-1, där n är ett heltal 

```python
for i in range(start, end, step):
    print(i)
``` 
printar i från start till end-1 med ett steg som bestämmer hur mycket den hoppar i varje iteration 

</div>

````python
for i in range(3):
    print(f"{i} Hej")
````

````python
for i in range(10):
    print(i, end=" ")
````

````python
sum_ = 0
for i in range(6): 
    sum_ += i

print(f"0+1+2+3+4+5={sum_}")
````

````python
sum_ = 0
for i in range(1, 20, 2):
    sum_ += i

    print(f"{i}", end="")

    # vid 19 så skrivs = summan istället för +     
    operator = "+" if i != 19 else f"={sum_}"

    print(operator, end="")
````

## random module

Used to generate random numbers. 

</div>

````python
import random as rnd # randommodulen kan användas med rnd 

num_six = 0
dice_rolls = 10000

# tärningssimulation
for i in range(dice_rolls):
    dice = rnd.randint(1,6) # genererar ett slumpmässigt tal mellan 1 och 6
    if dice == 6:
        num_six += 1

print(f"Antalet sexor {num_six} i {dice_rolls} tärningsrullningar")  
print(f"Sannolikheten att slå sex är {num_six/dice_rolls}")
````

## Loopa genom listor

En lista är en datastruktur som lagrar flera element i en variabel. Det kommer komma separat lektion om listor, men vi introducerar det här för att visa på for-loop över en lista. 

</div>


````python
fruits = ["apple", "orange", "grape", "strawberry"]

print("I love to eat")
for fruit in fruits: 
    print(fruit)
````

## pythonska loopar

- att skriva kod som följer idiomatisk python kallas för pythonskt (pythonic)

För loopar har vi 
- enumerate för att räkna upp en siffra samtidigt som vi går igenom en lista
- zip för att loopa igenom flera listor samtidigt

````python
print("My favorite fruits are: ")
# enumerate för att få siffra som räknas upp för varje element i listan
for i, fruit in enumerate(fruits,1): # 1 gives the starting number
    print(f"{i}. {fruit}")
````

loopa igenom flera listor samtidigt med zip

````python
fruits = ["apple", "orange", "grape", "strawberry"]
recipies = ["pie", "juice", "juice", "cake"]

print(f"{'FRUIT':<20}RECIPE")
for fruit, recipe in zip(fruits, recipies):
    print(f"{fruit:<20}{recipe}")
````

## Nästlad for loop

En for loop inuti en annan for loop. Den inre loopen kör för varje iteration i den yttre loopen.

</div>

````python
# Exempel: tic tac toe koordinator

for x in range(3):
    for j in range(3):
        print(f"({x},{j})", end="")    
    print() 
````
