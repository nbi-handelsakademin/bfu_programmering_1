# Matplotlib 

Python har en del bra verktyg för att hantera och analysera data för att få fram insikter. Ett sådant verktyg är matplotlib som är ett paket för att rita grafer i Python. Genom att rita grafer kan man utforska, förstå och hitta trender i datan.

</div>

## plt interface - procedurell

- använd inte denna approach
- det är bättre att använda oop-approachen som jag visar efter detta exempel

```python
import matplotlib.pyplot as plt

hours = range(5)
prices = [2, 1, 5, 3, -2]

plt.plot(hours, prices, 'o--')
plt.xlabel("timmar")
plt.ylabel("pris")
plt.title("Aktiepris för Cooltie")
```

## OOP approach 

- detta är mer pythonskt än den procedurella approachen med plt

```python
fig, ax = plt.subplots(1)
work_years = range(4)
salaries = [30_000, 32_000, 33_000, 35_000]

ax.plot(work_years, salaries, "--o")
ax.set(
    xlabel="Års erfarenhet",
    ylabel="Månadslön",
    title="Exempel på löneutveckling för en anställd",
)
```

## Exempel på bilpris


Detta är samma exempel vi använt i lektionen om while-satser

En ny elbil kostar 450 000 kr och förlorar ungefär 15 % av sitt värde varje år. Efter hur många hela år är den värd mindre än hälften av nypriset?

Här passar while-sats speciellt bra då vi inte i förväg vet hur många gånger vi behöver iterera

```python
values = [450_000]
current_value = values[0]
year = 0

while current_value > values[0] / 2:
    # tilldelningsoperator - gamla värdet multiplicerat med 0.85 och tilldelar till variabeln
    current_value *= 0.85
    values.append(current_value)
    year += 1

years = range(year + 1)
print(f"{years = }")
print(f"{values = }")

fig, ax = plt.subplots(1)
ax.plot(years, values, "--o")
ax.set(xlabel="År", ylabel="Värde (kr)", title="Värdeminskning på en ny elbil")
```

## Bar chart 

När vi har kategoriska värden så är det bättre med en bar chart, medans line chart är bättre för att visualisera förändring över tid

```python
off_days_2026 = {
    "jan": 11,
    "feb": 8,
    "mar": 9,
    "apr": 10,
    "maj": 12,
    "jun": 8,
    "jul": 8,
    "aug": 10,
    "sep": 8,
    "okt": 9,
    "nov": 9,
    "dec": 9,
}


fig, ax = plt.subplots(1, figsize=(12, 4))

ax.bar(x=off_days_2026.keys(), height=off_days_2026.values())
ax.set(
    xlabel="Månad",
    ylabel="Antal lediga dagar",
    title="Antal lediga dagar per månad 2025",
)
```
