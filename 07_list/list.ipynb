# list

lista är muterbar ordnad sekvens som kan lagra flera element av samma eller olika datatyper

## accessa element

- för listor använder man bracket operatorn [] för att plocka ut element
- kallas även indexering
- notera att Python räknar från noll

```py
l = [1,2,3]
l[0] # 1
l[1] # 2
```


````python
transportation = ["spårvagn", "tåg", "buss", "färja", "flygplan"]
transportation
````

````python
print(f"{transportation[0] = }")
print(f"{transportation[1] = }")
print(f"{transportation[2] = }")

# kan indexera från slutet med minus
print(f"{transportation[-1] = }")
print(f"{transportation[-3] = }")

# en sträng är en sekvens av karaktärer
print(f"{transportation[-3][0] = }")

# reversera en lista
print(f"{transportation[::-1] = }")
````

## Mutera en lista

en lista är muterbar, så vi kan ändra värden på listan

- byta ut ett värde på en index
- lägga till ett värde
- ta bort ett värde


````python
transportation[1] = "bil"
transportation.append("maglevtåg")
print(transportation)
````

````python
# tar bort elementet på index 3 och returnerar den
transportation.pop(3)
````

````python
transportation
````

````python
# sätta in värde på specifik index
transportation.insert(1, "jetplan")
transportation
````

````python
# notera dock att en sträng i sig inte är muterbar
transportation[1][1] = "a"
````

## Sammanslagning och repetition av lista

Går att använda + och \* operatorer på en list a

- `+` konkatenerar (slår ihop) två listor
- `*` repeterar en lista


````python
numbers = [3.14, 2.713, 1]
print(f"{numbers = }")
print(f"{numbers*2 = }")
print(f"{numbers*3 = }")
````

````python
numbers = [1, 2, 3]
letters = ["a", "b", "c"]

numbers + letters
````

## Slicing

- extrahera ut delar av sekvensen

````python
subjects = [
    "matematik",
    "svenska",
    "engelska",
    "programmering",
    "naturvetenskap",
    "samhällskunskap",
]

print(f"{subjects[2:] = }")
print(f"{subjects[2:4] = }")
print(f"{subjects[3:] = }")
````

## Iterera genom listan

vi kan loopa genom en lista med en for-sats


````python
for transport in transportation:
    print(transport)
````

````python
for i, transport in enumerate(transportation):
    print(f"{i}. {transport}")
````

## Lägga till värden med en for-loop


````python
squares = []

for i in range(-5, 5):
    squares.append(i**2)

squares
````

## list comprehension

Ett mer kompakt sätt att skapa lista i Python.


````python
squares = [x**2 for x in range(-5, 5)]
squares
````

````python
filtered_squares = [x**2 for x in range(-5, 5) if x**2 > 10]
filtered_squares
````

## Sortera lista

En lista kan sorteras på två sätt

sort() metod

- l.sort() -> muterar listan

sorted() funktion

- sorted(l) -> returnerar en ny lista

Sorteringen sker uppstigande ordning (ascending) by default. För att sortera nedstigande (descending), använd reverse=True


````python
greek_letters = ["delta", "beta", "gamma", "alpha"]
greek_letters.sort()
# returnerar None
````

````python
greek_letters
````

````python
numbers = [2,1,4,5]
print(f"{sorted(numbers) = }")
print("numbers ej muterad")
print(f"{numbers = }")
print(f"{sorted(numbers, reverse=True) = }")
````
