## Assignment operators

Assignment operator eller tilldelningsoperator är användbart för att uppdatera nuvarande värde på en variabel.

```python
x = 0 # tilldelar 0 till variabel x
x += 1 # x = x + 1 i.e. lägger till 1 till x
```

````python
# udda tal från -9 till 9

n = -9 
while n <= 10:
    print(n, end=" ") # detta ger ett mellanslag mellan varje tal
    n += 2 
````

## Exempel

<div style = "max-width:66ch;">

En ny elbil kostar 450 000 kr och förlorar ungefär 15 % av sitt värde varje år. Efter hur många hela år är den värd mindre än hälften av nypriset?

Här passar while-sats speciellt bra då vi inte i förväg vet hur många gånger vi behöver iterera

</div>

````python
start_price = 450_000
new_price = start_price
year = 0

while new_price > start_price / 2:
    # tilldelningsoperator - gamla värdet multiplicerat med 0.85 och tilldelar till variabeln
    new_price *= 0.85 
    year += 1
    # avrunda till 1000-tal och .0f formaterar till 0 decimaler
    print(f"År {year} är bilen värd {round(new_price, -3):.0f}kr")
````

## break-sats

med **break** kan vi avbryta loopar som while-loop.

````python
import random as rnd # this module has functions for generating random numbers 

while True: 
    number1 = rnd.randint(1,10)
    number2 = rnd.randint(1,10)

    user_answer = int(input(f"Vad blir {number1}*{number2}? "))
    if user_answer == number1*number2:
        print("Snyggt")
    else:
        print(f"{user_answer} är tyvärr fel")

    play_again = input("Spela igen? (J för ja) ")
    
    if play_again != "j":
        print("Ha en bra dag")
        break
````
