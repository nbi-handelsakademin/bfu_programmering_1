## Konkatenera strängar 

Slå ihop strängar med plusoperatorn `+`


</div>

```python
firstname = "Kokchun"
lastname = "Giang"

fullname = firstname + " " + lastname # concatenating strings
work = "Lärare"
address = "Supah cool streeeto"
phone = "3918401"

# multiline f-string
contact_details = f"""
Name: {fullname}
Work: {work}
Adress: {address}
Phone: {phone}
"""

print(contact_details)
```

## Konkatenera med loopar 

```python
bamba = ["Chili sin carne", "Nudlar", "Pokebowl", "Pannkakor", "Tacos"]
days = ["Må", "Ti", "On", "To", "Fr"]

menu = "Veckomeny\n" # \n gives newline

for day, food in zip(days, bamba):
    menu += f"{day}: {food} \n"

print(menu)
```

## Indexering

- använd indexering [] för att komma åt specifik element
- combine with slice operator : to get a section in the string

</div>

```python
# notera att det finns olika typer av quotes, har du använt dubbelcitat för strängen kan du inte använda det inuti strängen då 
# det skulle avsluta strängen
quote = "!False - it's funny because it's True" 
print(f"quote[0]: {quote[0]}")
print(f"quote[5]: {quote[5:10]}") # [start:slut-1]
print(f"quote[-4]: {quote[-4]}") # indexering start 4e positionen från slutet
print(f"quote[-4:]: {quote[-4:]}") # slice från -4 till slutet
print(f"Backwards quote[::-1]: {quote[::-1]}") # reversera
```

## Splitta och rensa strängar

- splittar sträng till lista och strängar

</div>


```python
subjects = "Math, biology  ,   cHeMistry  ,  PrOgramming      "
subjects = subjects.split(",") # specar vad den ska splitta på 
subjects = [subject.strip().upper() for subject in subjects] # strips and make uppercase

print(f"{subjects = }")

for subject in subjects: 
    print(f"{subject} has {len(subject)} letters") 
    # len() returnerar antalet element i sekvensen dvs antalet karaktärer
```

```python

numbers = input("Ange några tal separerade med mellanslag: ").split()
numbers = [float(number) for number in numbers]

sum = 0

for number in numbers:
    sum += number

mean = sum/len(numbers)
print(f"Medelvärdet av dina tal är {mean:.1f}")
```

## Regular expressions

Regex används för att söka i text mha olika mönster 
- findall(regex-pattern, string) returnerar alla matcher i en lista
- sub(regex-patter, string) för att ersätta ett mönster med ett värde
- split(regex-pattern, string) för att splitta baserat på ett mönster
- search(regex-pattern, string) index för första match

</div>

```python
import re 

# find all words ending with att in this text
text = "Det var en gång en katt som tog på sig en hatt."
# råsträng r för att inte ge specifik mening till ex backslash \ då \n betyder nyrad
# med råsträng så blir den ett faktiskt backslash
print(f"{re.findall(r".att", text) = }") # punkt . matchar alla karaktärer
```

```python
text = """Regex används för att söka i text mha olika mönster 
- findall(regex-ttern, string) returnerar alla matcher i en lista
- sub(regex-patter, string) för att ersätta ett mönster med ett värde
- split(regex-patern, string) för att splitta baserat på ett mönster
- search(regex-pattern, string) index för första match"""

# vill ersätta regex-patter och regex-pattern med regex_pattern
print(re.sub(r"[a-z]+-[a-z]+", "regex_pattern", text))
```
