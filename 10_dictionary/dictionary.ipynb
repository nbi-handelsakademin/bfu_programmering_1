# Lecture notes - dictionary

En datastruktur där datan lagras i form av key-value par. För att accessa ett värde i en dictionary används nyckeln

</div>

```python
# ett sätt att skapa dictionary med dict() funktionen
person = dict(
    name = "Kokchun",
    age = 35, 
    work = "lärare",
    interest = ["matte", "python", "schack"],
    phone = "031-9183 1321" # ring inte 
)

print(f"person dictionary: {person}")
print(f"Personens namn är {person['name']}") 
print(f"Personens ålder är {person['age']} år") 

print(f"{person['name']} tycker om {person['interest']}") 

for interest in person["interest"]:
    print(f"{person['name']} tycker om {interest}")
```

## Dictionary comprehension

Likt list comprehension som är ett kompakt sätt att loopa och skapa en lista finns dictionary comprehension som gör liknande för en dictionary
  
</div>


```python
import random as rnd 

# dict comprehension
grade_limit = {grade: i*10 for i, grade in enumerate("FEDCBA")}
print(f"Betygsgränser {grade_limit}")

scores = [rnd.randint(0,60) for _ in range(20)]

print(f"Poäng {scores}")

# count number in each grade category
grade_count = {key: 0 for key in grade_limit}

for score in scores:
    for grade, limit in grade_limit.items():
        if limit <= score < limit+10:
            grade_count[grade] += 1

print(f"Betygsfrekvens {grade_count}")
```
