# Funktioner 

Funktioner är kod som används för att återanvända kod istället för att upprepa koden. Detta följer en princip som kallas för DRY (Dont't Repeat Yourself)

</div>

<div style="max-width:66ch;">

## Skapa en funktion med def

```py
def name(param1, param2, ...): 
    statements
    ...

name(arg1, arg2, ...)
```


</div>

````python
def squarer(x): # input parameter x 
    return x**2 # returnerar x**2

print(squarer(3))
squares = [squarer(x) for x in range(10)]
print(squares)
````

## Defaultvärde

````python
# den här funktionen returnerar None då vi inte har specat en return
def say_hello(name = "friend"):
    print(f"Hi {name}") 

say_hello("Adam") # name blir nu Adam istället för friend
say_hello() # vi ger inget argument så name får defaultvärdet friend
print(say_hello("Beda")) # returnerar None och skriver ut det
````

````python
def smallest(number1, number2):
    if number1 > number2:
        return number2
    else:
        return number1

print(smallest(3,5))
````

## Keyword arguments

Keyword arguments gör att du kan skicka argument till en funktion med syntaxen funkt(parameternamn = värde). Det är användbart när en funktion har många parametrar och du vill göra funktionsanropet mer lättläst och självförklarande. Keyword arguments är också praktiska när du bara vill ange värden för vissa av parametrarna och inte vill vara beroende av argumentens ordning.

</div>


````python
def count_words(text):
    words = text.split()  # splittar på mellanslag
    return len(words)


quote = "I stand on the shoulders of giants"
print(
    count_words(text=quote)
)  # här använder vi keyword argument istället för positional
````

````python
# kan också använda med positional argument dvs baserat på dess position
count_words("hej hur mår du?")
````

## *args - variabelt antal argument

- du kan ange godtyckligt antal argument
- de sparas i en tuple

````python
def print_args(*args):
    print(type(args))
    print(args)

print_args(1,2,3)
````

````python
def largest_two(number1, number2, *numbers):
    numbers = [number1, number2] + list(numbers)
    numbers.sort(reverse=True)
    return numbers[0], numbers[1]


largest_two(2, 3, 4, -5, 1, 23)
````
