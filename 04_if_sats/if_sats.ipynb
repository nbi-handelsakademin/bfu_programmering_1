# If-satser

En if-sats används för att styra vad som ska hända i ett program baserat på om ett villkor är sant eller falskt

## if-else

````python
age = int(input("Enter your age:")) # 14
age_limit = 15

# note the syntax:
if age < age_limit:
    print("Din ålder är {age} år. Du behöver vara minst {age_limit} för att se den här videon")
    print("Titta på en tecknad film istället")
else:
    print("Välkommen till den här läskiga filmen")
````

### and operator

vänstra villkor `and` högra villkor

| Vänster    | Höger     | Resultat  |
| :--------  | :-------- | :-------- |
| **True**   | **True**  | **True**  |
| **True**   | **False** | **False** |
| **False**  | **True**  | **False** |
| **False**  | **False** | **False** |


````python
age = 9
height = 129

if age >= 7 and 130 <= height <= 200:
    print("Du kan åka balder")
else:
    print(
        "Tyvärr uppfyller du inte kraven minst 7 år och höjd mellan 130 och 200cm för att få åka"
    )

````

### or operator

vänstra villkor `or` högra villkor

| Vänster    | Höger     | Resultat  |
| :--------  | :-------- | :-------- |
| **True**   | **True**  | **True**  |
| **True**   | **False** | **True**  |
| **False**  | **True**  | **True**  |
| **False**  | **False** | **False** |


````python
is_weekend = True
is_sick = False 

if is_weekend or is_sick: 
    print("Stanna hemma från skolan")
else:
    print("Gå till skolan")
````

## if-elif-else

```py
if villkor: 
    satser 
    ...
elif villkor: 
    satser
    ...
...
elif villkor:
    satser
    ...

    satser

```

````python
points = 25

print("Ditt betyg för tentan är", end = " ")

if points > 60:
    print("A")
elif points > 50:
    print("B")
elif points > 40:
    print("C")
elif points > 30:
    print("D")
elif points > 20:
    print("E")
else:
    print("F")
````

## Modulooperatorn

modulooperatorn % beräknar resten vid en heltalsdivision

````python
print(f"{25 % 5 = }") # 25/5 = 5 rest 0
print(f"{25 % 4 = }") # 25/4 = 6 rest 1
print(f"{25 % 2 = }") # 25/2 = 12 rest 1 (ej jämnt delbart med 2 -> udda) 
````

### Nästlad if-sats
- if-sats i en if-sats
- går att nästla hur mycket man vill, men det gör koden svår att läsa

````python
cool_number = 5

if isinstance(cool_number, int):
    if cool_number % 2: # truthy -> finns rest -> udda
        print("udda")
    elif not cool_number % 2: # falsey -> finns ingen rest -> jämn
        print("jämnt") 
else:
    print("Udda och jämnt bara definerad för heltal")
````

## Truthy och Falsey

Värdet utvärderas till True (truthy) eller False (falsey)

Följande utvärderas till False och är falsey
- False
- None
- 0, -0, 0.0
- "" tom sträng
- [], (), {}, set() (tomma sekvenser)


````python
text = ""

# eftersom text är tom -> falsey, så gör not att falsey evalueras till True
if not text:
    print("du har inte skrivit något")
else: 
    print(f"du skrev: {text}")
````

````python
fruit_basket = ["äpplen", "bananer"]

if fruit_basket:
    print(f"Din fruktkorg innehåller {fruit_basket}")
````

## ternary if-sats

Struktur för en ternary if-sats eller en conditional expression. Vi kan skriva en if-sats på en rad 

```
[resultat om True] if [villkor] else [resultat om False]
```

````python
username = ""

username = username if username else "guest"

print(f"hi {username}")
````
