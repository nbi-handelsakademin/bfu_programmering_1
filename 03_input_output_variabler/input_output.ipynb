square example

````python
side = 4 
square_perimeter = side * 4
square_area = side**2 

print(f"Kvadratens sida är {side} l.e.")
print(f"Omkretsen av kvadraten är {square_perimeter} l.e")
print(f"Arean av kvadraten är {square_area} a.e")
````

````python
import math 

radius = 1
perimeter = 2*math.pi*radius
area = math.pi*radius**2

print(f"Cirkelns radie är {radius} l.e.")
print(f"Omkretsen av cirkeln är {perimeter} l.e")
print(f"Arean av cirkeln är {area} a.e")
````

````python
# format to 2 decimal
print(f"Omkretsen av cirkeln är {perimeter:.2f} l.e")
print(f"Arean av cirkeln är {area:.2f} a.e")
````

self documenting expressions with = operator in f-string

````python
name = "David"
radius = 2
gravity = 9.82
is_unit_circle = False


print(f"{name = }")
print(f"{radius = }")
print(f"{gravity = }")
print(f"{is_unit_circle = }")
````
